```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
texts = [
    "I am happy",
    "I am angry",
    "I feel sad",
    "I love this"
]
```

```python
labels = [
    "happy",
    "angry",
    "sad",
    "happy"
]
```

```python
print("=" * 60)
print("TRAINING DATA")
print("=" * 60)
```

```python
for sentence, emotion in zip(texts, labels):
    print(f"Text: {sentence} --> Emotion: {emotion}")
```

```python
vec = CountVectorizer()

# Learn vocabulary from training data.
# Convert text into numerical matrix.
X = vec.fit_transform(texts)
```

```python
print("\n")
print("=" * 60)
print("VOCABULARY")
print("=" * 60)

print(vec.get_feature_names_out())
```

```python
print("\n")
print("=" * 60)
print("VECTORIZED DATA")
print("=" * 60)

print(X.toarray())
```

```python
model = MultinomialNB()

# Train the model.
#
# X      -> Features
# labels -> Target classes
model.fit(X, labels)

print("\nModel Training Completed Successfully")
```

```python
new_text = ["sai is feeling alone"]

print("\n")
print("=" * 60)
print("TEST SENTENCE")
print("=" * 60)

print(new_text)
```

```python
test = vec.transform(new_text)


print("\nVectorized Test Data:")
print(test.toarray())
```

```python
prediction = model.predict(test)

# Display prediction.
print("\n")
print("=" * 60)
print("PREDICTED EMOTION")
print("=" * 60)

print(prediction[0])
```

```python
sample_sentences = [
    "I love my friends",
    "I am angry with you",
    "I feel very sad",
    "Today I am happy"
]

print("\n")
print("=" * 60)
print("MULTIPLE PREDICTIONS")
print("=" * 60)
```

```python
for sentence in sample_sentences:

    # Convert text into vector
    sample_vector = vec.transform([sentence])

    # Predict emotion
    result = model.predict(sample_vector)

    print(f"{sentence} --> {result[0]}")
```
