```python
import pandas as pd
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
url = "https://raw.githubusercontent.com/plotly/datasets/master/Wholesale%20customers%20data.csv"
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00292/Wholesale%20customers%20data.csv"
df = pd.read_csv(url)
```

```python
print("=" * 60)
print("FIRST FIVE RECORDS")
print("=" * 60)
print(df.head())
```

```python
print("\n")
print("=" * 60)
print("DATASET INFORMATION")
print("=" * 60)
print(df.info())
```

```python
print("\nDataset Shape:", df.shape)
```

```python
print("\nColumns Available:")
print(df.columns)
```

```python
X = df[
    [
        'Fresh',
        'Milk',
        'Grocery',
        'Frozen',
        'Detergents_Paper',
        'Delicassen'
    ]
]
```

```python
print("\n")
print("=" * 60)
print("SELECTED FEATURES")
print("=" * 60)
print(X.head())
```

```python
kmeans = KMeans(
    n_clusters=5,
    random_state=0
)
```

```python
df['Cluster'] = kmeans.fit_predict(X)
```

```python
print("\n")
print("=" * 60)
print("CUSTOMER CLUSTERS")
print("=" * 60)
```

```python
print(
    df[
        [
            'Channel',
            'Region',
            'Fresh',
            'Milk',
            'Cluster'
        ]
    ].head(20)
)
```

```python
print("\n")
print("=" * 60)
print("CLUSTER CENTERS")
print("=" * 60)

print(kmeans.cluster_centers_)
```

```python
sns.scatterplot(
    data=df,
    x='Fresh',
    y='Milk',
    hue='Cluster',
    palette='tab10',
    s=80
)
```

```python
plt.title(
    "Customer Personality Analysis using K-Means",
    fontsize=14
)
```

```python
plt.xlabel(
    "Annual Income (k$)"
)
```

```python
plt.ylabel(
    "Spending Score (1-100)"
)
```

```python
plt.grid(True)
```

```python
plt.tight_layout()
```

```python
plt.savefig("customer_segments.png")

# Display graph.
plt.show()
```
